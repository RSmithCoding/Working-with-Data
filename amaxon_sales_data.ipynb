```python
import pandas as pd

df = pd.read_csv("amazon_sales_data 2025.csv")

print(df)
```

```python
df = df[~df['Status'].str.contains("Cancelled", case=False, na=False)]
print(df)
```

```python
group = df.groupby("Customer Name")
print(group['Total Sales'].sum().sort_values(ascending=False))
```
